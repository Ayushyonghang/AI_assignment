```python
#flatten a nested list

def flatten(lst):
  flat_list = []
  for i in lst:
    for j in i:
      flat_list.append(j)
  return flat_list

nested_list = [[1,2],[3,4],[5,6]]
print("Flatten list: ", flatten(nested_list))
```

```python
#Accessing nested list elements

def access_nested_element(lst, indicies):
  element = lst
  for index in indicies:
    element = element[index]

  return element

list = [[1,2,3],[4,5,6],[7,8,9]]
indicies = [1,0]

print("Result: ", access_nested_element(list,indicies))
```

```python
def sum_nested(lst):
  sum = 0
  for num in lst:
    if isinstance(num,list):
      sum += sum_nested(num)

    elif isinstance(num, (int,float)):
      sum += num
  return sum

S_list = [[1,2],[3,[4,5]],6]
result = sum_nested(S_list)
print("The sum of elements is: ", result)
```

```python
#Remove specific element from a nested list
def remove_element(lst, elm):
    new_lst=[]
    for num in lst:
        if isinstance(num, list):
            sub_list = remove_element(num, elm)
            if sub_list:
                new_lst.append(sub_list)
        elif num != elm:
            new_lst.append(num)
    return new_lst

sample_list =  [[1, 2], [3, 2], [4, 5]]
element = 2
print("New List:", remove_element(sample_list, element))
```

```python
#find the maximum element in an nested list

def find_max(lst):
  max_num = 0
  for num in lst:
    if isinstance(num,list):
      max_num = max(max_num,find_max(num))
    else: max_num = max(max_num, num)
  return max_num

sample_list = [[1,2],[3,[4,7]],6]
print("The maximum number in the list is: ", find_max(sample_list))
```

```python
#Count occurences of an element in a nested list

def count(lst,element):
  n = 0
  for num in lst:
    if isinstance(num,list):
      n += count(num, element)
    elif num == element:
      n += 1
  return n

sample_list = [[1,2],[2,3],[2,4]]
element = 2
print(f"The number of {element} in the list is: ", count(sample_list,element))
```

```python
#Flatten a list of lists of lists

def flatten(lst):
  new_list = []
  for item in lst:
    if isinstance(item,list):
      new_list.extend(flatten(item))
    else:
      new_list.append(item)
  return new_list

sample_list = [[[1,2],[3,4]],[[5,6],[7,8]]]
print("New flattened list is: ", flatten(sample_list))
```

Task 8

```python
#Nested list average

def average_nested(lst):
  def helper(current_list):
    total = 0.0
    count = 0
    for element in current_list:
      if isinstance(element, list):
        sub_total, sub_count = helper(element)
        total += sub_total
        count += sub_count
      elif isinstance(element, (int,float)) and not isinstance(element,bool):
        total += element
        count += 1
    return total, count

  total_sum, total_count = helper(lst)
  return total_sum/total_count if total_count != 0 else 0.0

sample_list = [[1,2],[3,4],[5,6]]
print("The average is: ", average_nested(sample_list))


```

Problem-4: Matrix Operations

Task 1

```python
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])
```

```python
# A = np.array([[3,4],[7,8]]) and B = np.array([[5,3],[2,1]])
# 1. Prove A.A−1 = I

A_Inverse = np.linalg.inv(A)
I = np.eye(2)
result = np.dot(A,A_Inverse)
print("A.A-1: ",result)

print("\n Is A*A_Inverse an identity matrix? ", np.allclose(result,I))
```

Task 2

```python
# Prove AB not equal to BA

AB = np.dot(A,B)
BA = np.dot(B,A)
print("\n Ab = \n", AB, "\n")
print("BA = \n", BA, "\n")

print("Is AB equal to BA? ", np.array_equal(AB,BA))
```

Task 3

```python
# Prove (AB)^T = B^T * A^T
AB_T = np.transpose(AB)
BT_AT = np.dot(np.transpose(B),np.transpose(A))
print("\n (AB)^T =\n", AB_T, "\n")
print("B^T * A^T =\n", BT_AT, "\n ")
print("(AB)^T = B^T * A^T ? ", np.array_equal(AB_T, BT_AT))
```

Element wise Multiplication
• Using Python Lists, perform element-wise multiplication of two lists of size 1, 000, 000.
Measure and Print the time taken for this operation.

• Using Numpy Arrays, Repeat the calculation and measure and print the time taken for
this operation.

```python
import time
def list_multiplication(list1, list2):
  result = []
  for i in range (len(list1)):
    result.append(list1[i]*list2[i])
  return result

def array_multiplication(arr1, arr2):
  return arr1*arr2

size = 1000000

list1 = list(range(size))
list2 = list(range(size))
list_start_time = time.time()
list_result = list_multiplication(list1, list2)
list_end_time = time.time()
list_time = list_end_time - list_start_time

array1 = np.arrange(size)
array2 = np.arrange(size)
arr_start_time = time.time()
arr_result = array_multiplication(array1, array2)
arr_end_time = time.time()
arr_time = arr_end_time - arr_start_time

print(f"Time taken by list: {list_time:'6f'} seconds")
print(f"Time taken by array: {arr_time:.6f} seconds")
```

Dot Product
• Using Python Lists, compute the dot product of two lists of size 1, 000, 000. Measure and
Print the time taken for this operation.

• Using Numpy Arrays, Repeat the calculation and measure and print the time taken for
this operation.

```python
def list_dotproduct(list1, list2):
  dp_list = 0
  for i in range (len(list1)):
    dp_list += list1[i]*list2[i]
  return dp_list

def array_dotproduct(arr1, arr2):
  return np.dot(arr1, arr2)

size = 1000000

list1 = list(range(size))
list2 = list(range(size))
list_start_time = time.time()
list_result = list_dotproduct(list1, list2)
list_end_time = time.time()
list_time = list_end_time - list_start_time

array1 = np.arrange(size)
array2 = np.arrange(size)
arr_start_time = time.time()
arr_result = array_dotproduct(array1, array2)
arr_end_time = time.time()
arr_time = arr_end_time - arr_start_time

print(f"Time taken by list: {list_time:'6f'} seconds")
print(f"Time taken by array: {arr_time:.6f} seconds")
```
